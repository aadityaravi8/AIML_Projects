```python
result3_df = pd.read_csv('result_3.csv')
column_order = ['Very Bad', 'Bad', 'Okay', 'Good', 'Very Good']
result3_df['CUSTOMER_FEEDBACK'] = pd.Categorical(result3_df['CUSTOMER_FEEDBACK'], categories=column_order, ordered=True)
result3_df
```

```python
pivot_df = result3_df.pivot(index='QUARTER_NUMBER', columns='CUSTOMER_FEEDBACK', values='QUARTERLY_PER_FEEDBACK_PERCENTAGE')
pivot_df = pivot_df.rename_axis('Quarter Number', axis='index').rename_axis('Customer Feedback', axis='columns')

pivot_df
```

```python
colors = ['red', 'lightcoral', 'yellow', 'lightgreen', 'green']

pivot_df.plot(kind='bar', stacked=True, figsize=(10, 4), color=colors)
plt.title('Trend of Customer Satisfaction')
plt.xlabel('Quarter Number')
plt.ylabel('Percentage split of customer feedback')
plt.xticks(rotation=360)

legend_labels = ['Very Bad', 'Bad', 'Okay', 'Good', 'Very Good']
legend_handles = [plt.Rectangle((0, 0), 1, 1, color=color) for color in colors]
plt.legend(legend_handles, legend_labels, loc='upper right', bbox_to_anchor=(1.2, 1));
```

```python
result7_df = pd.read_csv('result_7.csv')
result7_df.head()
result7_df['QoQ_REVENUE_CHANGE_PERCENTAGE'] = result7_df['QoQ_REVENUE_CHANGE_PERCENTAGE'].replace(np.NaN, 0)
```

```python
plt.figure(figsize=(8, 3))
sns.lineplot(data=result7_df, x='QUARTER_NUMBER', y='QoQ_REVENUE_CHANGE_PERCENTAGE', marker='o', color='blue')
plt.title('Quarter on Quarter % change in Revenue')
plt.xlabel('QUARTER NUMBER')
plt.ylabel('QoQ REVENUE CHANGE PERCENTAGE')
plt.grid(True)
plt.xticks(result6_df['QUARTER_NUMBER'])
#plt.yticks(result6_df['ORDER_COUNT'])
plt.show()
```

```python
result8_df = pd.read_csv('result_8.csv')
result8_df['REVENUE'] = result8_df['REVENUE'].apply(lambda x: "{:.2f}".format(x))
result8_df.head()
```

```python
plt.figure(figsize=(8, 3))
sns.lineplot(data=result7_df, x='QUARTER_NUMBER', y='REVENUE', marker='o', color='blue')
plt.title('Quarter on Quarter % change in Revenue')
plt.xlabel('QUARTER NUMBER')
plt.ylabel('QoQ REVENUE CHANGE PERCENTAGE')
plt.grid(True)
plt.xticks(result6_df['QUARTER_NUMBER'])
#plt.yticks(result6_df['ORDER_COUNT'])
plt.show()
```

```python
sns.set_style("whitegrid")

fig, ax1 = plt.subplots(figsize=(8, 4))

# Plot Revenue on the first y-axis
sns.lineplot(data=result8_df, x='QUARTER_NUMBER', y='REVENUE', color='blue', marker='o', label='REVENUE', ax=ax1)

# Create the second axis and plot Order Count on it
ax2 = ax1.twinx()
sns.lineplot(data=result8_df, x='QUARTER_NUMBER', y='ORDER_COUNT', color='green', marker='s', label='ORDER COUNT', ax=ax2)

# Set labels and title
ax1.set_xlabel('Quarter Number')
ax1.set_ylabel('Revenue', color='b')
ax2.set_ylabel('Order Count', color='g')
plt.title('Trend of Revenue and Orders by Quarter')
plt.xticks(result8_df['QUARTER_NUMBER'])
plt.yticks(result8_df['ORDER_COUNT'])

# Display the legend
ax1.legend(loc='center left')
ax2.legend(loc='center right')

# Show the plot
plt.show()
```
