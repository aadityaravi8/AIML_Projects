##### 2. An electrical manufacturing company conducts quality checks at specified periods on the products it manufactures. Historically, the failure rate for the manufactured item is 5%. Suppose a random sample of 10 manufactured items is selected. Answer the following questions. [4 Marks]

```python
import numpy as np
import scipy.stats as stats
n=10 #Sample size
p=0.05 #Failure rate of the manufactured item
k=np.arange(0,11) #Array of quantiles
binomial = stats.binom.pmf(k,n,p)
binomial
```

##### 2.A. Probability that none of the items are defective? [1 Mark]

```python
print("Probability that none of the items are defective: ",binomial[0])
```

##### 2.B. Probability that exactly one of the items is defective? [1 Mark]

```python
print("Probability that exactly one of the items is defective: ",binomial[1])
```

##### 2.C. Probability that two or fewer of the items are defective? [1 Mark]

```python
binomial_c = stats.binom.cdf(k,n,p)
print("Probability that two or fewer of the items are defective: ",binomial_c[2])
```

##### 2.D. Probability that three or more of the items are defective ? [1 Mark]

```python
print("Probability that three or more of the items are defective: ",1-binomial_c[2])
```

##### 3.  A car salesman sells on an average 3 cars per week.

```python
rate = 3
n=np.arange(0,17)
```

##### 3.A. What is Probability that in a given week he will sell some cars? [1 Mark]

```python
#cumulative poisson distribution
poisson_c = stats.poisson.cdf(n,rate)
#Probability that in a given week he will sell some cars
p=1-poisson_c[0]
print("Probability that in a given week he will sell some cars: ",p)
```

##### 3.B. What is Probability that in a given week he will sell 2 or more but less than 5 cars? [1 Mark]

```python
#Probability that he sells less than 5 cars i.e upto 4 cars
P1 = poisson_c[4]
#Probability that he sells less than 2 cars i.e up to 1 car
P2 = poisson_c[1]
#Probability that in a given week he will sell 2 or more but less than 5 cars
P=P1-P2
print("Probability that in a given week he will sell 2 or more but less than 5 cars: ",P)
```

##### 3.C. Plot the poisson distribution function for cumulative probability of cars sold per-week vs number of cars sold per week. [1 Mark]

```python
import matplotlib.pyplot as plt
%matplotlib inline

plt.plot(n,poisson_c, 'o-')
plt.title('Poisson Distribution')
plt.xlabel('Number of cars sold per week')
plt.ylabel('Cumulative Probability of cars sold per week')
plt.show()
```

#### Problem Statement

A company ABC every year provides AWS Solution Architect certification training to its 50 employees on first come first serve basis and gives the vouchure to appear in the exam at free of cost. The company has been doing this from last 7 years. The success rate to pass the certification exam is 60%. 
#### Answer the following question?
Find the probility that none of the employee passes the exam?  
Find the probability that exactly 20 employees pass the exam?  
Find the probability that 25 or more passes the exam?

The above real life problem statement is based on Probability Distribution and Binomial Distribution

```python
n=50 #sample size
p=0.6 #probability to pass the certification exam based on historic data as given
k=np.arange(0,51) #array of quantiles
binomial=stats.binom.pmf(k,n,p) #probability mass function

#Find the probility that none of the employee passes the exam
print("Probility that none of the employee passes the exam: %.5f" %binomial[0])

#Find the probability that exactly 20 employees pass the exam
print("Probability that exactly 20 employees pass the exam: %.5f" %binomial[20])

#Find the probability that 25 or more passes the exam
binomial_c=stats.binom.cdf(k,n,p) #cumulative density function
print("Probability that 25 or more passes the exam: %.5f" %(1-binomial_c[24]))

# Let us plot the distribution for number of successes and its probability
plt.bar(k, binomial)
plt.title("Binomial Distribution: n=%i , p=%.2f" % (n, p))
plt.xlabel("Number of Successes") 
plt.ylabel("Probability of Successes")
plt.show() 
```

##### A. Is there any significant difference between Funds raised by companies that are still operating vs companies that closed down? [1 Mark]

#### We will do hypothesis testing to get answer of this question, let us visualize this a bit for better understanding

```python
startup_filter = startup.loc[(startup['OperatingState'].isin(['Operating','Closed']))]
plt.figure(figsize=(15,5))
sns.boxplot(data=startup_filter,x='OperatingState',y='Funds_in_million')
plt.show()
```

* All the startup which were closed were less funded compared to operating ones but yes it is also important to mention that few startups have similar Funding as closed ones, but they are still operating

#### Let us to Hypothesis testing to identify the answer of this question

* 2 independent samples are given
    * first thoses startups which are closed
    * second those startups which are operating
    
#### Hence it would be a 2 sample independent ttest

* H0: "There is no significant difference between Funds raised by companies that are still operating vs companies that closed down"
* HA: "There is significant difference between Funds raised by companies that are still operating vs companies that closed down"

```python
import scipy.stats as stats
#Operating Startup Funds_in_million sample
OStartupFundsInMillion = startup.loc[startup['OperatingState']=='Operating']['Funds_in_million']
#Closed Startup Funds_in_million sample
CStartupFundsInMillion = startup.loc[startup['OperatingState']=='Closed']['Funds_in_million']
#Consider level of significance as 0.05
alpha=0.05
#Do 2 sample independent ttest
t_statistic, p_value = stats.ttest_ind(OStartupFundsInMillion, CStartupFundsInMillion, equal_var=True)
print('t_statistic:', t_statistic, "\np_value:",p_value)

if(p_value<alpha):
    print("There is significant difference between Funds raised by companies that are still operating vs companies that closed down")
else:
    print("There is no significant difference between Funds raised by companies that are still operating vs companies that closed down")
    
```

##### B. Write the null hypothesis and alternative hypothesis. [1 Mark]

#### Null Hypothesis:
**H0:** "There is no significant difference between Funds raised by companies that are still operating vs companies that closed down"
#### Alternate Hypothesis:
**HA:** "There is significant difference between Funds raised by companies that are still operating vs companies that closed down"

##### C. Test for significance and conclusion [1 Mark]

```python
import scipy.stats as stats
#Operating Startup Funds_in_million sample
OStartupFundsInMillion = startup.loc[startup['OperatingState']=='Operating']['Funds_in_million']
#Closed Startup Funds_in_million sample
CStartupFundsInMillion = startup.loc[startup['OperatingState']=='Closed']['Funds_in_million']
#Consider level of significance as 0.05
alpha=0.05
#Do 2 sample independent ttest
t_statistic, p_value = stats.ttest_ind(OStartupFundsInMillion, CStartupFundsInMillion, equal_var=True)
print('t_statistic:', t_statistic, "\np_value:",p_value)

if(p_value<alpha):
    print("There is significant difference between Funds raised by companies that are still operating vs companies that closed down")
else:
    print("There is no significant difference between Funds raised by companies that are still operating vs companies that closed down")
    
```

#### Conclusion:
* Since p_value is greater than alpha i.e level of significance, we fail to reject the null hypothesis which means There is **no** significant difference between funds raised by companies that are still operating vs companies that closed down.
